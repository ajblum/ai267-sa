```python
import os
from model_registry import ModelRegistry
```

```python
INTERNAL_REGISTRY_URL = "http://redhattraining-model-registry.rhoai-model-registries.svc.cluster.local:8080"

print(f"Initializing connection to internal service: {INTERNAL_REGISTRY_URL}")

# Initialize the official client library
# Note: is_secure=False because your internal service listens on plain HTTP port 8080
registry = ModelRegistry(
    server_address=INTERNAL_REGISTRY_URL, 
    is_secure=False, 
    author="AI267-developer-user"
)
```

```python
def register_new_lab_model(model_name, version_str, s3_uri, format_type="onnx"):
    print(f"\nProcessing entry: {model_name} ({version_str})...")
    try:
        # Atomic registration wrapper
        model = registry.register_model(
            name=model_name,
            version=version_str,
            uri=s3_uri,
            model_format_name=format_type,
            model_format_version="1"
        )
        print(f"✅ Success! Assigned DB ID: {model.id}")
        return model
    except Exception as e:
        print(f"❌ Failed to register model. Reason: {e}")
        return None
```

```python
models_to_import = [
    {"name": "diabetes-1", "version": "v2.25.1", "uri": "oci://registry.ocp4.example.com:8443/redhattraining/models/diabetes:2.25.1"},
    {"name": "diabetes-2", "version": "v2.25.2", "uri": "oci://registry.ocp4.example.com:8443/redhattraining/models/diabetes:2.25.2"},
    {"name": "diabetes-3", "version": "v2.25.3", "uri": "oci://registry.ocp4.example.com:8443/redhattraining/models/diabetes:2.25.3"},
    {"name": "diabetes-4", "version": "v2.25.4", "uri": "oci://registry.ocp4.example.com:8443/redhattraining/models/diabetes:2.25.4"},
    {"name": "diabetes-5", "version": "v2.25.5", "uri": "oci://registry.ocp4.example.com:8443/redhattraining/models/diabetes:2.25.5"}

]

for item in models_to_import:
    register_new_lab_model(item["name"], item["version"], item["uri"])
```
